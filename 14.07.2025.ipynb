```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])

print("Array:\n", arr)
print("Dimensions (ndim):", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Sum of elements:", np.sum(arr))
print("Mean of elements:", np.mean(arr))
print("Sorted array:\n", np.sort(arr))
print("Sine of elements (in radians):\n", np.sin(arr))
```

```python
from numpy.linalg import det, eig

matrix = np.array([[4, 2], [1, 3]])

print("Matrix:\n", matrix)
print("Determinant:", det(matrix))

eigenvalues, eigenvectors = eig(matrix)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

```python
one_d_list = [i for i in range(1, 13)]  # list with 12 elements
array_1d = np.array(one_d_list)

array_2d = array_1d.reshape(3, 4)
array_3d = array_1d.reshape(2, 2, 3)

print("Original 1D array:", array_1d)
print("Reshaped to 2D (3x4):\n", array_2d)
print("Reshaped to 3D (2x2x3):\n", array_3d)
```

```python
zeros_array = np.zeros((2, 3))
ones_array = np.ones((3, 3))
identity_matrix = np.eye(4)
random_matrix = np.random.rand(2, 2)
range_array = np.arange(1, 10)

print("Zeros array:\n", zeros_array)
print("Ones array:\n", ones_array)
print("Identity matrix:\n", identity_matrix)
print("Random matrix:\n", random_matrix)
print("Range array:\n", range_array)
```

```python
from scipy import linalg

matrix2 = np.array([[6, 1], [4, -2]])
det_value = linalg.det(matrix2)

print("Matrix:\n", matrix2)
print("Determinant using scipy:", det_value)
```

```python
eigvals, eigvecs = linalg.eig(matrix2)

print("Eigenvalues using scipy:", eigvals)
print("Eigenvectors using scipy:\n", eigvecs)
```
